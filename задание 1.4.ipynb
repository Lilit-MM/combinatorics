```python
import sympy as sp

# Объявляем переменные
x = sp.Symbol('x', real=True)
C = sp.Symbol('C', real=True)

# 1. Находим константу С из уравнения нормировки
integral_eq = sp.integrate(C * x**(-3/2), (x, 1, sp.oo))
C_solved = sp.solve(sp.Eq(integral_eq, 1), C)[0]
print(f"Постоянная C: {C_solved:.2f}")

# 2. Находим вероятность на интервале (5, 10)
density = C_solved * x**(-3/2)
prob_integral = sp.integrate(density, (x, 5, 10))

# Выводим округленный ответ
print(f"Округленный ответ: {prob_integral.evalf(4)}") 
```
