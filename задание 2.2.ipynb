```python
import sympy as sp

# 1. Объявляем переменные
r = sp.Symbol('r', positive=True)
R = sp.Symbol('R', positive=True)

# 2. Задаем плотность распределения f(r) = 2r / R^2 (это соотношение попадания точки в круг, который явным образом будет меньше, чем круг радиусом 3).
# это и будет вероятность
f = (2 * r) / (R**2)

# 3. Находим математическое ожидание M[ξ] = ∫ r * f(r) dr от 0 до R
M_xi_expr = sp.integrate(r * f, (r, 0, R))

# 4. Находим второй начальный момент M[ξ^2] = ∫ r^2 * f(r) dr от 0 до R
M_xi2_expr = sp.integrate(r**2 * f, (r, 0, R))

# 5. Вычисляем дисперсию D[ξ] = M[ξ^2] - (M[ξ])^2
D_xi_expr = M_xi2_expr - (M_xi_expr)**2

# Подставляем значение R = 3
R_value = 3
M_xi = M_xi_expr.subs(R, R_value)
D_xi = D_xi_expr.subs(R, R_value)

print(f"Формула матожидания: {M_xi_expr} => При R=3: {M_xi}")
print(f"Формула дисперсии: {D_xi_expr} => При R=3: {D_xi}")
```
