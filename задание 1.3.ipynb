Брошены три игральные кости. Какова вероятность, что на одной из них выпала
единица, если известно, что выпали разные числа?

```python
# найти вероятность события (A) (выпала единица) при условии, что произошло событие (B) (выпали разные числа). Нужно найти 
# P(A∣B)= P(A∩B) / P(B)
```

```python
P_B = (6*5*4)/6**3
P_A_and_B = 3*4*5/6**3
P_B_given_A = P_A_and_B / P_B
print(P_B_given_A)
```

Из колоды, состоящей из 36 карт, наугад выбирают две карты. С какой вероятностью они окажутся одной масти? Ответ округлить до четверого знака после запятой.

```python
import math

# Всего способов вытащить 2 карты из 36
total_ways = math.comb(36, 2)

# Способов вытащить 2 карты одной конкретной масти (из 9 карт)
ways_per_suit = math.comb(9, 2)

# Так как мастей всего 4, умножаем на 4
favorable_ways = 4 * ways_per_suit

# Расчет вероятности
probability = favorable_ways / total_ways

print(f"Всего исходов: {total_ways}")        
print(f"Благоприятных: {favorable_ways}")   
print(f"Ответ:         {probability:.4f}")   
```
